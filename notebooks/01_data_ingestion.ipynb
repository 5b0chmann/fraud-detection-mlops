```python
import shutil
shutil.copy("/kaggle/input/creditcardfraud/creditcard.csv","data/raw/creditcard.csv")
```

```python
df = pd.read_csv("data/raw/creditcard.csv")
print(df.shape)
df.head()
```

```python
print("Shape:")
print(df.shape)

print("\nFehlende Werte:")
print(df.isnull().sum().sum())

print("\nKlassenverteilung:")
print(df["Class"].value_counts())
```

```python
fraud_rate = df["Class"].mean() * 100
print(f"Betrugsrate: {fraud_rate:.4f}%")
```

```python
import json

report = {"rows": int(df.shape[0]),
           "columns": int(df.shape[1]),
           "missing_values": int(df.isnull().sum().sum()),
           "fraud_cases": int(df["Class"].sum()),
           "normal_cases": int((df["Class"] == 0).sum())}
print(report)
```

```python
with open("reports/data_validation.json", "w") as f:
  json.dump(report, f, indent=4)
```
